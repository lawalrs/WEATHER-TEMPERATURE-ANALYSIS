```python
# HANDS-ON 2 Python Lists vs NumPy Arrays
# Task 1 — Try It With Python Lists
sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]

result = sales_timothy + sales_brenda

print(result)

# 1. Did Python produce: [15, 25, 35]?
# 2. What did Python actually do?
# 3. Why do you think that happened?

# 1. No, Python did not produce [15, 25, 35].
# 2. Python joined (concatenated) the two lists together.
# 3. This happened because the + operator for Python lists
# means concatenation, not element-by-element addition.
```

```python
# Task 2 — Convert Them to NumPy Arrays
sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]

timothy_array = np.array(sales_timothy)
brenda_array = np.array(sales_brenda)
combined_sales = timothy_array + brenda_array
print(combined_sales)
```

```python
# Task 3 — Compare the Results
sales_timothy = [10, 15, 20]
sales_brenda = [5, 10, 15]

# Python lists
list_result = sales_timothy + sales_brenda

# NumPy arrays
array_result = np.array(sales_timothy) + np.array(sales_brenda)

print("Python lists:", list_result)
print("NumPy arrays:", array_result)

# NumPy is useful because it allows us to perform
# mathematical operations on data element by element.
# This makes it easier to calculate, compare, and
# manipulate numerical data in an organized way.
```

```python
# Challenge 1
# Which location recorded the highest temperature on each day?

# Temperature data
lagos = np.array([27.5, 29.0, 30.5, 28.0, 31.0])
abuja = np.array([25.0, 28.0, 29.5, 30.0, 32.0])
ibadan = np.array([26.5, 28.5, 29.0, 27.5, 30.5])

# Combine the data into a 2D array
temperatures = np.array([
    lagos,
    abuja,
    ibadan
])

# Location names
locations = np.array(["Lagos", "Abuja", "Ibadan"])

# Find the location with the highest temperature for each day
highest = np.argmax(temperatures, axis=0)

# Display the result
for day in range(5):
    location = locations[highest[day]]
    temperature = temperatures[highest[day], day]

    print(f"Day {day + 1}: {location} - {temperature}°C")
```

# Explain why NumPy becomes increasingly useful as the amount of numerical data grows.

NumPy becomes increasingly useful as the amount of numerical data
grows because it allows large amounts of numerical data to be stored
and processed using arrays.

Instead of manually processing each value, NumPy allows us to perform
mathematical operations on entire arrays at once. It also provides
functions for calculating averages, minimums, maximums, comparisons,
and other numerical operations.

Therefore, as a dataset becomes larger, NumPy makes numerical analysis
more convenient, organized, and efficient.

```python
# 3. Import NumPy
import numpy as np
```

```python
# 4. Dataset

## Dataset

# The weather station recorded temperatures over five days.

# The locations are:

# - Lagos
# - Abuja
# - Ibadan

# All temperatures are measured in degrees Celsius (°C).

lagos = [27.5, 29.0, 30.5, 28.0, 31.0]

abuja = [25.0, 28.0, 29.5, 30.0, 32.0]

ibadan = [26.5, 28.5, 29.0, 27.5, 30.5]

print("Lagos:", lagos)
print("Abuja:", abuja)
print("Ibadan:", ibadan)
```

```python
# 5. Convert Data to NumPy Arrays

lagos = np.array(lagos)
abuja = np.array(abuja)
ibadan = np.array(ibadan)

print("Lagos:", lagos)
print("Abuja:", abuja)
print("Ibadan:", ibadan)

# Now the three datasets are NumPy arrays.
```

```python
# 6. Array Inspection

print("Lagos dimensions:", lagos.ndim)
print("Lagos shape:", lagos.shape)
print("Lagos data type:", lagos.dtype)

print("Abuja dimensions:", abuja.ndim)
print("Abuja shape:", abuja.shape)
print("Abuja data type:", abuja.dtype)

print("Ibadan dimensions:", ibadan.ndim)
print("Ibadan shape:", ibadan.shape)
print("Ibadan data type:", ibadan.dtype)

# This means each location has 5 temperature values stored in a 1D NumPy array.
```

```python
# 7. Numerical Operations
# Challenge 2 — Increase every temperature by 1°C

lagos_plus_1 = lagos + 1
abuja_plus_1 = abuja + 1
ibadan_plus_1 = ibadan + 1

print("Lagos + 1°C:", lagos_plus_1)
print("Abuja + 1°C:", abuja_plus_1)
print("Ibadan + 1°C:", ibadan_plus_1)
```

```python
# 8. Lagos vs Abuja Comparison
## Lagos vs Abuja Comparison

#T he Lagos and Abuja temperature arrays were compared by subtracting
# Abuja's temperature from Lagos's temperature.

# A positive result means Lagos was warmer than Abuja, while a negative
# result means Lagos was cooler than Abuja.

difference = lagos - abuja

print("Lagos vs Abuja temperature difference:")
print(difference)
```
